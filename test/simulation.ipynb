```julia
using TensorCrossInterpolation, LinearAlgebra
using QuanticsTCI
using QuanticsGrids
using ITensorMPS
using ITensors
using NDTensors
using TCIITensorConversion
using IterativeSolvers
using LinearMaps
using LinearAlgebra
using HDF5
include("../src/2D_QIS.jl")
```

```julia
# set simulation parameters

R = 10
max_bond = 39
dx = 1 / (2^R-1)
delta_t = 0.1 * 2.0^-(R-1)
nu = 1e-5
penalty_coefficient = 2.5e5


# setup quantics grid and sites

grid = DiscretizedGrid{2}(R, (0,0), (1,1); includeendpoint = true)
s = siteinds("Qudit", R, dim=4)

println("del_t = $(trunc(delta_t*1000, digits=4)) ms")
```

```julia
# setup initial velocity field

x_min = 0.4
x_max = 0.6
h = 1/200

function d_1_func(x, y)
    out = 2 / h^2 * ( (y - x_max) * exp( -(y - x_max)^2 / h^2 ) + (y - x_min) * exp( -(y - x_min)^2 / h^2 ) )
    return out * ( sin(8π * x) + sin(24π * x) + sin(6π * x) ) 
end

function d_2_func(x, y)
    out = π * ( exp( -(y - x_max)^2 / h^2 ) + exp( -(y - x_min)^2 / h^2 ) )
    return out * ( 8*cos(8π * x) + 24*cos(24π * x) + 6*cos(6π * x) ) 
end

function A_func(x, y)
    return sqrt( d_1_func(x,y)^2 + d_2_func(x,y)^2 )
end

A = find_max_on_2D_grid(A_func, R)
u_0 = 1.
delta_var = u_0 / (40 * A)

function J_1_func(y)
    return u_0 / 2 * ( tanh( (y - x_min)/h ) - tanh( (y - x_max)/h ) - 1 )
end

function D_1_func(x, y)
    return delta_var * d_1_func(x, y)
end

function D_2_func(x, y)
    return delta_var * d_2_func(x, y)
end


# initial velocity field is sum of a jet profile and a disturbance

function u_init_1_func(x, y)
    return J_1_func(y) + D_1_func(x, y)
end

function u_init_2_func(x, y)
    return D_2_func(x, y)
end

print("Initial velocity field defined. ")
```

```julia
 # build mps with QuanticsTCI

u1q, ranks1, errors1 = quanticscrossinterpolate(Float64, u_init_1_func, grid; maxbonddim=max_bond)
u2q, ranks2, errors2 = quanticscrossinterpolate(Float64, u_init_2_func, grid; maxbonddim=max_bond)


# convert to ITensorMPS format

ux = ITensorMPS.MPS(TensorTrain(u1q.tci), sites=s)
uy = ITensorMPS.MPS(TensorTrain(u2q.tci), sites=s)


# build base MPO's

d1x = Diff_1_8_x(dx, s)
d1y = Diff_1_8_y(dx, s)
d2x = Diff_2_8_x(dx, s)
d2y = Diff_2_8_y(dx, s)
dx_dx = apply(d1x, d1x, maxdim=max_bond)
dy_dy = apply(d1y, d1y, maxdim=max_bond)
dx_dy = apply(d1x, d1y, maxdim=max_bond)
del = MPO([delta(s[i], s[i]', s[i]'') for i in 1:length(s)])
print("Base MPO's built. ")
```

```julia
# Start time evolution and save results in hdf5 file

f = h5open("velocityfieldsMPS_chi_$(max_bond)_v2.h5","w")
write(f,"ux_t_0",ux)
write(f,"uy_t_0",uy)

n_evo = 500 # number of evolution steps
save_every = 10 # save every n evolution steps

@assert n_evo % save_every == 0 "n_evo must be a multiple of save_every"

t0 = time() # measure time entire simulation

for i in 1:n_evo
    ux, uy = RK4_step(ux, uy, delta_t, nu, d1x, d1y, d2x, d2y, del, dx_dx, dy_dy, dx_dy, max_bond)
    if i % save_every == 0
        write(f,"ux_t_$i",ux)
        write(f,"uy_t_$i",uy)
    end
end

t_total = time()-t0

write(f,"total_simulation_time",t_total)
close(f)
```
