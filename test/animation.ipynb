```julia
using HDF5
include("../src/visualization.jl")
plotlyjs()
```

```julia
# open HDF5 file and read total simulation time and grid size from the data
# make sure to run the simulation before trying to visualize the data

file_name = "velocityfieldsMPS_chi_39_v1.h5"
data = h5open(file_name, "r")

mps = read(data, "/uy_t_0", MPS)
R = length(mps)
delta_t = 0.1 * 2.0^-(R-1)

T = read(data, "total_simulation_time")
m, sek = Int(div(T,60)), trunc(T%60, digits=4)
h, m = div(m, 60), m % 60
d, h = div(h, 24), h % 24
print("$d d | $h h | $m min | $sek sek")
```

```julia
# save snapshot of the velocity field at time step i as pdf

i = 500

mps = read(data, "/uy_t_$i", MPS)
plot_mps(mps, grid_size=10, title="t = $(trunc(i * delta_t * 1e3, digits=1)) ms", save_name="figure.pdf")
```

```julia
# create animation of the velocity field evolution over time

anim = @animate for i in 0:500
    mps = read(data, "/uy_t_$i", MPS)
    plot_mps(mps, grid_size=8, title="t = $(trunc(i * delta_t * 1e3, digits=1)) ms")
end

gif(anim, "uy_t.gif", fps=30)
```

```julia
# close file

close(data)
```
